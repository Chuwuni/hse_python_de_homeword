```python
import json
from itertools import islice
```

```python
purchases = {}
with open(file='/home/ivan/Downloads/purchase_log.txt', mode='r', encoding='utf-8') as file:
    for line in file:
        line = line.strip()
        if line:
            data = json.loads(line)
            purchases[data['user_id']] = data['category']
```

```python
for key, value in islice(purchases.items(), 3):
    print(f"{key}: {value}")
```

```python
with open(file="purchase_log.json", mode='w', encoding='utf-8') as write_file:
    json.dump(purchases, write_file, indent=4, ensure_ascii=False)
```
