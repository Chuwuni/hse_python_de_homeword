## Задание 1

```python
example_date = [
    "The Moscow Times - Wednesday, October 2, 2002",
    "The Guardian - Friday, 11.10.13",
    "Daily News - Thursday, 18 August 1977"
]
```

```python
formate_dict = {
    "The Moscow Times": "%A, %B %d, %Y",
    "The Guardian": "%A, %d.%m.%y",
    "Daily News": "%A, %d %B %Y"
}
```

```python
example_date = list(map(lambda x: x.split(" - "), example_date))
```

```python
for i in example_date:
    print(f"{i[0]}: {datetime.strptime(i[1], formate_dict.get(i[0]))}")
```

```python
for i in example_date:
    print(f"{datetime.strptime(i[1], formate_dict.get(i[0]))}")
```

## Задание 2

```python
def date_range(start_date: str, end_date: str):
    try:
        start_date = datetime.strptime(start_date, "%Y-%m-%d")
        end_date = datetime.strptime(end_date, "%Y-%m-%d")
        result_list = []
        if start_date > end_date:
            return result_list
        else:
            curr_date = start_date
            while curr_date <= end_date:
                result_list.append(curr_date.strftime("%Y-%m-%d"))
                curr_date+=timedelta(days=1)
            return result_list
    except ValueError:
        return []
            
```

```python
print(date_range("2022-01-01", "2022-01-03"))
```

```python
print(date_range("2022-01-03", "2022-01-01"))
```

```python
print(date_range("2022-02-30", "2022-02-31"))
```
